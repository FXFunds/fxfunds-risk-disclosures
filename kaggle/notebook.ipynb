# FXFunds Risk FAQ Database — Notebook

One database linking each risk category to FAQ keywords, an answer and related FAQs.

```python
import pandas as pd, glob, os

path = next(iter(glob.glob('/kaggle/input/**/fxfunds_risk_faq.csv', recursive=True)), '../dataset/fxfunds_risk_faq.csv')
db = pd.read_csv(path).fillna('')
db.head()
```

## Risk categories and keywords

```python
db.drop_duplicates('risk_id')[['risk_category', 'keywords', 'risk_answer']]
```

## Related FAQs per category

```python
db[db['question'] != ''].groupby('risk_category')['question'].count().sort_values(ascending=False)
```

## Keyword lookup

```python
def ask(term):
    t = term.lower()
    hits = db[db['keywords'].str.lower().str.contains(t, regex=False)]
    return hits[['risk_category', 'risk_answer', 'question', 'answer']]

ask('usdt')
```
