# FXFunds Risk Disclosures — Document Register

Which official FXFunds risk disclosures and legal documents have been located, which are missing, and which documents to check for each type of risk.

- Website: https://fxfunds.com/
- GitHub: https://github.com/FXFunds/fxfunds-risk-disclosures

*Informational only; not legal or financial advice.*

```python
import glob
import pandas as pd

# Finds the CSVs wherever Kaggle mounts the dataset
def find(name):
    return glob.glob(f'/kaggle/input/**/{name}', recursive=True)[0]

reg = pd.read_csv(find('fxfunds_document_register.csv'))
risks = pd.read_csv(find('fxfunds_risk_categories.csv'))
reg[['id','category','document','status_label','last_checked']]
```

## Status of official documents

```python
reg['status_label'].value_counts()
```

## Status by category

```python
reg.groupby(['category','status']).size().unstack(fill_value=0)
```

## Documents not yet located

```python
reg[reg['status'] != 'located'][['id','category','document','should_cover']]
```

## Risk categories and the documents to check

```python
pairs = risks.assign(register_id=risks['register_ids'].str.split(';')).explode('register_id')
pairs = pairs.merge(reg[['id','document','status']], left_on='register_id', right_on='id', suffixes=('_risk',''))
pairs[['name','register_id','document','status']]
```

## Risks whose documents are all missing

```python
covered = pairs.groupby('name')['status'].apply(lambda s: (s == 'located').any())
covered[~covered].index.tolist()
```
